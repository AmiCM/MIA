### Ejercicio 2 · Una lista dentro de otra

Resuelve el ejercicio en la celda siguiente y conserva las comprobaciones.

```python
lote_a = ["hola", "gracias"]
lote_b = ["adiós", "hasta luego"]
con_append = lote_a.copy()
con_extend = lote_a.copy()

# Aplica append y extend por separado e inspecciona los resultados.
# Inicio de los ejemplos de append y extend.
resultado_append = con_append.append(lote_b) # agrega como otro elemento
resultado_extend = con_extend.extend(lote_b) # agrega cada sub-elemento

print("append:", con_append, len(con_append))
print("extend:", con_extend, len(con_extend))

# Lista plana con los cuatro textos y los dos últimos mediante un corte.
plana = con_extend
ultimos_dos = plana[-2:]
print(f"Dos últimos: {ultimos_dos}")
```

**Predicción:** `append(lote_b)` deja una lista de **3** elementos cuyo último elemento es

4 textos. Ambos métodos devuelven `None`.

**Explicación:** `append` siempre añade un único objeto, sea cual sea su tipo; `extend`
recorre el iterable recibido y añade cada uno de sus elementos. El corte `[-2:]`

### Mutabilidad, alias y copia

Dos nombres pueden apuntar a la misma lista. Para modificar el contenedor por separado, debemos copiarlo.
Si contiene otros objetos mutables, una copia superficial aún puede compartirlos. Este detalle explica muchos errores silenciosos
en pipelines de datos.

```python
original = ["dato-a", "dato-b"]
alias = original
copia = original.copy()

alias.append("dato-c")
copia.append("dato-independiente")

print("original:", original)
print("alias:   ", alias)
print("copia:   ", copia)
print("¿Mismo objeto?", original is alias, original is copia)
```

### Copiar una estructura anidada

`.copy()` hace una **copia superficial**: crea el contenedor externo, pero conserva
las referencias a sus elementos. Si dentro hay una lista, esa lista sigue compartida.

```python
registro_original = {"texto": "hola", "etiquetas": ["saludo"]}
registro_copia = registro_original.copy()
registro_copia["etiquetas"].append("corto")
print(registro_original)
print(registro_original is registro_copia)
print(registro_original["etiquetas"] is registro_copia["etiquetas"])
```

Crea una copia que permita añadir `corto` sin cambiar las etiquetas del original. Para esta estructura basta copiar también la lista interna. **Profundización:** investiga qué cambiaría si hubiera más niveles.

```python
fuente = {"texto": "hola", "etiquetas": ["saludo"]}
# Construye una copia independiente para esta estructura.

# Copia superficial del diccionario + copia de la lista interna.
copia_independiente = fuente.copy()
copia_independiente["etiquetas"] = fuente["etiquetas"].copy()
copia_independiente["etiquetas"].append("corto")

print("fuente:", fuente)
print("copia: ", copia_independiente)
```

**Explicación:** `.copy()` crea un diccionario nuevo pero conserva las referencias a sus

### Diccionarios: representar registros

Un diccionario asocia claves a valores. `in` comprueba las claves; `.items()`
permite recorrer pares clave–valor. `get` usa su valor por defecto cuando la clave

### Conjuntos: pertenencia, coincidencias y diferencias

Un conjunto conserva valores únicos. Usa `set()` para el conjunto vacío: `{}` crea

No uses la posición ni el orden impreso de un conjunto como parte del resultado.

```python
permitidas = {"positivo", "neutral", "negativo"}
observadas = {"positivo", "positivo", "desconocido"}
print("Reconocidas:", observadas & permitidas)
print("No reconocidas:", observadas - permitidas)
```

### Bloques, ciclos y decisiones

Los dos puntos abren un bloque y la sangría indica qué instrucciones pertenecen a

### Ejercicio 5 · Cero no es ausencia

Recorre `[None, 0.0, 0.4, 0.95, 0.99]`. Ignora `None` y detente en el primer score mayor o igual a 0.9. Deben mostrarse 0.0, 0.4 y 0.95. ¿Qué se perdería usando `if not score`?

Resuelve el ejercicio en la celda siguiente y conserva las comprobaciones.

```python
valores_busqueda = [None, 0.0, 0.4, 0.95, 0.99]
# Usa for, continue y break

mostrados = []
for score in valores_busqueda:
    if score is None:
        continue
    print(score)
    mostrados.append(score)
    if score >= 0.9:
        break

print(mostrados)

# Comparación con "if not score"
mostrados_con_not = []
for score in valores_busqueda:
    if not score:
        continue
    mostrados_con_not.append(score)
    if score >= 0.9:
        break
print(mostrados_con_not)
```

**Predicción:** se muestran `0.0`, `0.4` y `0.95`; `0.99` nunca se visita porque `break`
termina el ciclo en `0.95`.

**Explicación:** `continue` salta `None` y `break` corta en el primer score `>= 0.9`.
Con `if not score` se perdería el `0.0`, porque cero es falso en una condición y se
confundiría con un dato ausente. `score is None` pregunta exactamente por la ausencia.

## 4. Ciclos y comprehensions

Un `for` recorre los elementos de un iterable. En una lista puedes recorrer los
valores directamente; no hace falta construir sus índices. Usa `enumerate` cuando
también necesites la posición y `zip` para avanzar por varias secuencias a la vez.
Por defecto, `zip` se detiene al agotarse la secuencia más corta. Si necesitas
comprobar que tengan la misma longitud, usa `zip(..., strict=True)`.

```python
textos = ["muy bueno", "regular", "excelente"]
scores = [0.82, 0.51, 0.96]

for posicion, (texto, score) in enumerate(zip(textos, scores), start=1):
    print(f"{posicion}. {texto:<12} → {score:.0%}")
```

Las comprehensions sirven para transformaciones pequeñas y legibles. Si la
lógica requiere varias decisiones o efectos secundarios, un ciclo explícito
suele ser mejor.

```python
scores_altos = [score for score in scores if score >= 0.80]
longitudes = {texto: len(texto) for texto in textos}
iniciales = {texto[0] for texto in textos}

print(scores_altos)
print(longitudes)
print(iniciales)
```

### Ejercicio 6 · Transformar y filtrar

Convierte a Fahrenheit las temperaturas mayores o iguales a 25. Usa la fórmula `fahrenheit = celsius * 9 / 5 + 32`. Primero usa un ciclo y después una comprehension. El resultado debe ser `[77.0, 84.2, 87.8]`.

Resuelve el ejercicio en la celda siguiente y conserva las comprobaciones.

```python
temperaturas = [18, 21, 25, 29, 31]

# Con ciclo
fahrenheit_ciclo = []
for celsius in temperaturas:
    if celsius >= 25:
        fahrenheit_ciclo.append(celsius * 9 / 5 + 32)
print(fahrenheit_ciclo)

# Con comprehension.
fahrenheit_comprehension = [
    celsius * 9 / 5 + 32 for celsius in temperaturas if celsius >= 25
]
print(fahrenheit_comprehension)
```

**Explicación:** ambas versiones filtran (`>= 25`, incluye el 25) y transforman con la misma
fórmula. La comprehension expresa «transforma y filtra» en una línea legible; el ciclo es
preferible cuando la lógica crece (varias decisiones, mensajes, acumuladores).

## 5. Funciones: nombrar una transformación

Una función bien enfocada reduce repetición y permite probar una idea de forma
aislada. Sus parámetros son entradas; `return` produce una salida. Una función
sin `return` explícito devuelve `None`.

```python
def normalizar_texto(texto, *, minusculas=True):
    '''Elimina espacios externos y opcionalmente convierte a minúsculas.'''
    resultado = texto.strip()
    if minusculas:
        resultado = resultado.lower()
    return resultado


print(normalizar_texto("  Hola, IA  "))
print(normalizar_texto("  Hola, IA  ", minusculas=False))
```

El `*` vuelve explícitos los argumentos posteriores: debemos escribir
`minusculas=False`. En la semana 2 añadiremos anotaciones de tipo y Pydantic;
hoy nos concentramos en el comportamiento.

### Verificar una función con `assert`

```python
assert normalizar_texto("  HOLA  ") == "hola"
assert normalizar_texto("  HOLA  ", minusculas=False) == "HOLA"
assert normalizar_texto("") == ""
```

### Ejercicio 7 · Promedio sin inventar datos

Escribe `promedio_o_none(valores)`: devuelve el promedio o `None` si la lista está vacía. Comprueba `[0.0]`, `[]` y `[0.5, 1.0]`. ¿Por qué cero y ausencia necesitan resultados distintos?

Resuelve el ejercicio en la celda siguiente y conserva las comprobaciones.

```python
# Define promedio_o_none y escribe tres comprobaciones.
def promedio_o_none(valores):
    if not valores:
        return None
    return sum(valores) / len(valores)


print(promedio_o_none([0.0]) == 0.0)
print(promedio_o_none([0.0]) is not None)
print(promedio_o_none([]) is None)
print(promedio_o_none([0.5, 1.0]) == 0.75)
```

**Explicación:** `[0.0]` contiene una medición cuyo promedio es cero; `[]` no contiene
mediciones. Si ambos devolvieran `0`, quien use la función no podría distinguir «el
promedio fue cero» de «no había datos», y un reporte podría inventar un promedio.
Devolver `None` hace visible la ausencia. Aquí `if not valores` es correcto porque
evalúa la **lista** (vacía o no), no el valor numérico. Además evita un
`ZeroDivisionError` al dividir entre `len([])`.

### Ordenar usando una función

`sorted` crea una lista nueva; `.sort()` modifica la lista existente. El argumento
`key` recibe una función que calcula el criterio de orden. Pasamos la función
sin llamarla: Python la aplicará a cada elemento.

```python
evaluaciones = [
    {"modelo": "a", "score": 0.7},
    {"modelo": "b", "score": 0.95},
    {"modelo": "c", "score": 0.8},
]

def obtener_score(evaluacion):
    return evaluacion["score"]

ranking = sorted(evaluaciones, key=obtener_score, reverse=True)
print(ranking)
# lambda permite expresar una función pequeña en el mismo lugar.
assert ranking == sorted(evaluaciones, key=lambda e: e["score"], reverse=True)
```

### De contar a agrupar

Un contador responde «¿cuántos?». Una agrupación conserva los elementos asociados
a cada clave. `defaultdict(list)` crea una lista al acceder a una clave ausente.

```python
from collections import defaultdict

conteo_manual = {}
for etiqueta in etiquetas:
    conteo_manual[etiqueta] = conteo_manual.get(etiqueta, 0) + 1
assert conteo_manual == dict(Counter(etiquetas))

por_etiqueta = defaultdict(list)
for registro in [
    {"etiqueta": "positivo", "score": 0.9},
    {"etiqueta": "neutral", "score": 0.6},
    {"etiqueta": "positivo", "score": 0.8},
]:
    por_etiqueta[registro["etiqueta"]].append(registro["score"])
print(dict(por_etiqueta))
```

### Ejercicio 8 · Resumir grupos

Usando `por_etiqueta`, construye un diccionario con el promedio de cada grupo. Comprueba que positivo tiene dos scores y neutral uno. **Profundización:** ordena las etiquetas por promedio descendente.

Resuelve el ejercicio en la celda siguiente y conserva las comprobaciones.

```python
# Recorre por_etiqueta.items() y calcula cada promedio.

promedios_por_etiqueta = {}
for etiqueta, scores_grupo in por_etiqueta.items():
    promedios_por_etiqueta[etiqueta] = sum(scores_grupo) / len(scores_grupo)

print(promedios_por_etiqueta)

# Etiquetas por promedio descendente.
orden_por_promedio = sorted(
    promedios_por_etiqueta, key=promedios_por_etiqueta.get, reverse=True
)
print(orden_por_promedio)
```

**Explicación:** `por_etiqueta` conserva los scores de cada grupo, por lo que el promedio se
calcula con `sum / len` por grupo. Ningún grupo está vacío (`defaultdict` sólo crea una
clave cuando se le agrega un score), así que no hay división entre cero. Los floats se
comparan con `math.isclose` porque `(0.9 + 0.8) / 2` no es exactamente `0.85` en binario.
Para ordenar, `sorted` recorre las claves y `key=promedios_por_etiqueta.get` usa el
promedio de cada una como criterio.

## 8. POO: objetos que conservan estado

Una **clase** define cómo construir objetos y qué operaciones ofrecen. Cada
**instancia** es un objeto concreto. Sus **atributos** guardan datos y sus
**métodos** son funciones asociadas al objeto.

`__init__` inicializa una instancia; `self` representa esa instancia dentro de los
métodos. Al llamar `contador.agregar()`, Python pasa `contador` como `self`.

```python
class Contador:
    def __init__(self, nombre):
        self.nombre = nombre
        self.total = 0

    def agregar(self):
        self.total += 1

    def reiniciar(self):
        self.total = 0

contador_a = Contador("lote A")
contador_b = Contador("lote B")
contador_a.agregar()
contador_a.agregar()
print(contador_a.nombre, contador_a.total)
print(contador_b.nombre, contador_b.total)
assert contador_b.total == 0
```

### Laboratorio POO: una lista compartida por accidente

```python
class RegistroCompartido:
    registros = []  # Defecto intencional: la lista pertenece a la clase.

    def agregar(self, texto):
        self.registros.append(texto)

compartido_a = RegistroCompartido()
compartido_b = RegistroCompartido()
compartido_a.agregar("dato de A")
print(compartido_b.registros)
assert compartido_a.registros is compartido_b.registros
```

## 9. Caso integrador: un registro de predicciones

La clase siguiente conserva predicciones ya normalizadas y combina estructuras de
datos con comportamiento y estado. Cada registro recibe su posición al agregarse.
`resumen` devuelve cantidad, conteos y promedio; el promedio es `None` cuando no
hay registros.


```python
predicciones = [
    {"texto": "Excelente", "etiqueta": "positivo", "score": 0.98},
    {"texto": "Regular", "etiqueta": "neutral", "score": 0.66},
    {"texto": "Llegó roto", "etiqueta": "negativo", "score": 0.93},
    {"texto": "Cero válido", "etiqueta": "neutral", "score": 0.0},
]
```

```python
class RegistroPredicciones:
    def __init__(self, nombre):
        self.nombre = nombre
        self._registros = []

    def agregar(self, registro, posicion):
        guardado = registro.copy()
        guardado["posicion"] = posicion
        self._registros.append(guardado)

    def resumen(self):
        scores = [registro["score"] for registro in self._registros]
        return {
            "cantidad": len(self._registros),
            "conteos": dict(Counter(r["etiqueta"] for r in self._registros)),
            "promedio": sum(scores) / len(scores) if scores else None,
        }

lote = RegistroPredicciones("Demostración")
for posicion, registro in enumerate(predicciones):
    lote.agregar(registro, posicion)
print(lote.resumen())
```

### Leer el resultado antes de extenderlo

Deben quedar **4 registros**, conteos de positivo: 1, neutral: 2 y negativo: 1,
y promedio **0.6425**. Este promedio describe los scores; no mide exactitud del
modelo porque no tenemos etiquetas de referencia.

### 1 y 2 · Promedio por grupo y `reiniciar()`

Definimos una clase nueva con el mismo contrato que `RegistroPredicciones` y dos métodos
más.

```python
from collections import Counter, defaultdict


class RegistroPrediccionesAmpliado:
    def __init__(self, nombre):
        self.nombre = nombre
        self._registros = []

    def agregar(self, registro, posicion):
        guardado = registro.copy()
        guardado["posicion"] = posicion
        self._registros.append(guardado)

    def resumen(self):
        scores = [registro["score"] for registro in self._registros]
        return {
            "cantidad": len(self._registros),
            "conteos": dict(Counter(r["etiqueta"] for r in self._registros)),
            "promedio": sum(scores) / len(scores) if scores else None,
        }

    def promedio_por_etiqueta(self): # nuevo
        grupos = defaultdict(list)
        for registro in self._registros:
            grupos[registro["etiqueta"]].append(registro["score"])
        return {etiqueta: sum(s) / len(s) for etiqueta, s in grupos.items()}

    def reiniciar(self): # nuevo
        self._registros = []


# Comprobación de comportamiento de la clase ampliada.
ampliado_a = RegistroPrediccionesAmpliado("Lote A")
ampliado_b = RegistroPrediccionesAmpliado("Lote B")
for posicion, registro in enumerate(predicciones):
    ampliado_a.agregar(registro, posicion)
    ampliado_b.agregar(registro, posicion)

promedios_a = ampliado_a.promedio_por_etiqueta()
print(promedios_a)

ampliado_a.reiniciar()
print(ampliado_a.resumen())
print(ampliado_b.resumen())
```

**Explicación:** `promedio_por_etiqueta` agrupa con `defaultdict(list)`, igual que en el
ejercicio 8. Cada grupo tiene al menos un score, así que no hay división entre cero. El
lote vacío devuelve `{}`, no promedios inventados. El neutral `0.0` se conserva y baja
el promedio del grupo a `0.33`.

`reiniciar` asigna una **lista nueva** a `self._registros`, así que sólo cambia el
atributo de esa instancia. `ampliado_b` conserva sus 4 registros. La notebook habla de
«válidas y rechazos», pero esta clase sólo guarda registros válidos ya normalizados; si
tuviera una lista de rechazos, `reiniciar` también tendría que asignarle una lista nueva.


## Respuestas del cierre

**Elección de estructuras.**
- `list`: Me interesa entender el orden y cambiarlo a voluntad.
- `tuple`: Igual el orden pero sin cambiarlo, como nombres o coordenadas.
- `dict`: Un registro con campos con nombre, por ejemplo identificar IDs con nombres propios o de estados.
- `set`: Elementos diferentes sin orden especifico.

**Cero frente a ausencia.** `0.0` Es un valor medido en sí mismo, mientras que `None` indica que no hay valor (vacío). Usualmente se distinguen con `valor is None` (no con `if not valor`), y una función como `promedio_o_none([])` devuelve `None` en lugar de inventar un cero, dependiendo el caso de uso.

**Copiar contenedores con elementos mutables.** El método `.copy()` es superficial: El contenedor es externo es nuevo, pero las listas o diccionarios internos siguen compartidos. Hay que copiar también los niveles internos que se modificarán o de lo contrario tendrán la misma asignación de memoria.

**`RegistroPredicciones`.**
- Atributos: `nombre` y `_registros` (lista interna; el `_` indica uso interno).
- Métodos: `__init__` (inicializa el estado), `agregar(registro, posicion)` (guarda una copia con su posición) y `resumen()` (cantidad, conteos y promedio o `None`).
- Dos lotes deben tener listas distintas porque representan conjuntos de datos independientes, pero si compartieran la lista (como `RegistroCompartido`), agregar a un lote alteraría el resumen del otro. Por eso la lista se crea en `__init__`. Hay que tener cuidado con las asignaciones de memoria para evitar errores de ejecución o fallas de seguridad.
